```python
## Time Complexity O(log n)
## Space complecity O(log n) -- there no extra space is using , but when the recursive call happen, call will store in stack memory everytime
## Function Definition
def BinarySearch(arr,i,j,x):
    while i<=j:
        mid = i + (j-i)//2
        if arr[mid] == x:
            return mid
        elif arr[mid] < x:
            return BinarySearch(arr,mid+1,j,x) # Or use normal way instead of recurtion ---> i = mid+1
        else:
            return BinarySearch(arr,i,mid-1,x) # Or use normal way instead of recurtion  ---> j = mid
    return -1

## Driver Code
arr = [1,2,3,6,7,9,13,17,20,23,34,50]
x = 9
i = 0
j = len(arr)-1
## Function Call
result = BinarySearch(arr,i,j,x)
print(result)
```

```python
## Time Complexity : O(log n)
## Space Complexity : O(1) -- no recursion stack used

def BinarySearch(arr,i,j,x):
    while i <= j:
        mid = i + (j - i) // 2
        if arr[mid] == x:
            return mid
        elif arr[mid] < x:
            i = mid + 1
        else:
            j = mid - 1
    return -1
## Driver Code
arr = [1,2,3,6,7,9,13,17,20,23,34,50]
x = 17
i = 0
j = len(arr) - 1
result = BinarySearch(arr,i,j,x)
print(result)
```
