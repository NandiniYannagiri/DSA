```python
def fractionalKnapsack(profits,weights,M):
    fractions = [0]*len(profits)
    objects = list(range(len(profits)))
    ratio = [profit/weight for profit,weight in zip(profits,weights)]
    objects.sort(key=lambda i: ratio[i],reverse=True)
    # to take the objects sequentially, decreasing the weight & increasing the profit
    max_profit = 0
    for i in objects:
        if weights[i]<=M: 
            M-=weights[i]
            max_profit+=profits[i]
            fractions[i] = 1
        else:
            max_profit+=profits[i]*(M/weights[i])
            fractions[i]=M/weights[i]
            break
    return (max_profit,fractions)
weights = [5, 10, 12, 4, 7, 9, 3]
profits = [25,75,100,50,45,90,30]
M = 37
max_profit,fractons = fractionalKnapsack(profits,weights,M)
print(max_profit,fractons)
    
```
