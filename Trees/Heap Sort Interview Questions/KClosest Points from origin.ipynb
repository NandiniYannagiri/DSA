```python
from heapq import heappush, heappop
from math import sqrt

def get_distance(x, y):
    return sqrt(x**2 + y**2)
def k_closest(points, k):
    min_heap = []
    for x, y in points:
        heappush(min_heap, (get_distance(x, y), [x, y]))
    result = []
    for i in range(k):
        result.append(heappop(min_heap))
    return result

points = [[3,3],[5,-1],[-2,4]]
k = 2
result = k_closest(points, k)
print(result)
```
