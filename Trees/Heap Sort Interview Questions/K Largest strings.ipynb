```python
import heapq
def k_largest_strings(strings, k):
    min_heap = []
    for s in strings:
        heapq.heappush(min_heap, (len(s), s))
        if len(min_heap) > k:
            heapq.heappop(min_heap)
    # Extract only strings
    result = [item[1] for item in min_heap]
    return result
strings = ['python', 'programming', 'to', 'welcome']
k=2
Ls = k_largest_strings(strings,k)
print(Ls)
```

```python
import heapq
def k_largest_strings(strings, k):
    return sorted(heapq.nlargest(k, strings), reverse=True)
strings = ["apple", "banana", "kiwi", "strawberry", "grape"]
k = 3
print(k_largest_strings(strings, k))
```
