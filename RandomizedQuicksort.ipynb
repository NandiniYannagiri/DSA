```python
## Implementation of RandomizedQuickSort
## Recurrence relation T(n) = T(m-p) + T(q-m)
## Time complexity for best case Scenario T(n) = T(n/2)+T(n/2)+n --> O(nlogn)
## Time complexity for worst case Scenario T(n) = T(n-1)+n --> O(n^2)
## Space Complexity ---> O(1)
## Function definition for Randomized pivot element
## pivot element is randomly selecting using random function
import random
def RandomPartition(arr,p,q):
    ## Random pivot
    random_pivot = random.randrange(p,q)
    ## swap is happening
    arr[p],arr[random_pivot]=arr[random_pivot],arr[p]
    ## Function Call
    return Partition(arr,p,q)
## Function Definition for Partition Alogorithm --> O(n)
def Partition(arr,p,q):
    i=p
    ## starting element as pivot element
    pivot = arr[p]
    for j in range(i+1,len(arr)):
        if arr[j]<=pivot:
            i = i+1
            arr[i],arr[j]=arr[j],arr[i]
    ## Final swap between arr[i] and pivot element
    arr[i],arr[p]=arr[p],arr[i]
    return i
## Function Definition
def QuickSort(arr,p,q):
    if p<q:
        ## Function Calling for partition alogorithm
        m = Partition(arr,p,q)
        ## Recursive call for left side array
        ## T(m-p)
        QuickSort(arr,p,m-1)
        ## Recursive call for right side array
        ## T(q-m)
        QuickSort(arr,m+1,q)
    return arr
    
arr = [12,100,4,2,5,7,43,1,3,30,9,90]
p = 0
q = len(arr)-1
## Function Call
result = QuickSort(arr,p,q)
print(result)
```
