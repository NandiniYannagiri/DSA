```python
class Node:
    def __init__(self,data):
        self.data = data
        self.left = None
        self.right =  None
def is_FBT(root):
    if root is None:
        return True
    if root.left is None and root.right is None:
        return True
    if root.left and root.right:
        return is_FBT(root.left) and is_FBT(root.right)
    return False
root = Node(1)
root.left = Node(2)
root.right = Node(3)
root.right.left = Node(4)
is_FBT(root)
        
        
```
