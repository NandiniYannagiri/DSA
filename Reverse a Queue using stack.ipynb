```python
# Reverse a Queue - using stack
from collections import deque
def reverse_queue(q):
    stack = []
    # Move queue elements to stack
    while q:
        stack.append(q.popleft())
    # Move back to queue
    while stack:
        q.append(stack.pop())
# Driver
q = deque([10, 20, 30])
print("Original:", list(q))
reverse_queue(q)
print("Reversed:", list(q))
```
