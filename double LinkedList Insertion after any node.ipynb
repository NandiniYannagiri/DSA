```python
# doubly linked list after any node
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None
        self.prev = None
class DoublyLinkedList:
    def __init__(self):
        self.head = None
    # Insertion at End
    def InsertionAtEnd(self, new_data):
        new_node = Node(new_data)
        if self.head is None:
            self.head = new_node
            return
        temp = self.head
        while temp.next:
            temp = temp.next
        temp.next = new_node
        new_node.prev = temp
    # Print List
    def printList(self):
        temp = self.head
        while temp:
            print(temp.data, end=" ")
            temp = temp.next
        print()
    # Insert After Any Node
    def InsertionAfterAnyNode(self, prev_node, new_data):
        if prev_node is None:
            print("Previous node cannot be None")
            return
        new_node = Node(new_data)
        # Step 1
        new_node.next = prev_node.next
        new_node.prev = prev_node
        # Step 2 (important for DLL)
        if prev_node.next is not None:
            prev_node.next.prev = new_node
        # Step 3
        prev_node.next = new_node
```

```python
dll = DoublyLinkedList()

dll.InsertionAtEnd(2)
dll.InsertionAtEnd(4)
dll.InsertionAtEnd(5)
dll.InsertionAtEnd(10)

dll.printList()

dll.InsertionAfterAnyNode(dll.head, 12)
dll.printList()

dll.InsertionAfterAnyNode(dll.head.next, 1)

dll.printList()
```

```python
# double linked list insertion after any node using position
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None
        self.prev = None
class DoublyLinkedList:
    def __init__(self):
        self.head = None
    def insert_at_position(self, data, pos):
        new_node = Node(data)
        # Case 1: Insert at beginning
        if pos == 0:
            if self.head is not None:
                self.head.prev = new_node
                new_node.next = self.head
            self.head = new_node
            return
        temp = self.head
        count = 0
        # Traverse to position-1
        while temp is not None and count < pos - 1:
            temp = temp.next
            count += 1
        # If position is invalid
        if temp is None:
            print("Invalid Position")
            return
        # Case 2 & 3: Middle or End
        new_node.next = temp.next
        new_node.prev = temp
        if temp.next is not None:
            temp.next.prev = new_node
        temp.next = new_node
    def display(self):
        temp = self.head
        while temp:
            print(temp.data, end=" ")
            temp = temp.next
```

```python
dll = DoublyLinkedList()

dll.insert_at_position(10, 0)
dll.insert_at_position(20, 1)
# dll.insert_at_position(30, 2)
# dll.insert_at_position(15, 1)   # insert in middle

dll.display()
```
