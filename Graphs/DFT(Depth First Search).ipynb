```python
# Depth First Traversal Algorithm
def DFT(visited,graph,node):
    if node not in visited:
        print(node,end=" ")
        visited.add(node)
        for adjacentNodes in graph[node]:
            if adjacentNodes not in visited:
                # Recursion --> stack is used to store the function call
                DFT(visited,graph,adjacentNodes)
# Driver Code
visited = set()
# storing the data of graph using adjacency list
# Directed Graph
graph = { 
    '5' : ['3','7'],
    '3' : ['2','4'],
    '7' : ['8'],
    '4' : ['8'],
    '2' : [],
    '8' : []
}
DFT(visited,graph,'5')
```
