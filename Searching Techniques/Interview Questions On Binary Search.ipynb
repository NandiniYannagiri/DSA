```python
# Find the infinity index in an array
def BinarySearchElement(arr1,i,j,x):
    if i<=j:
        mid = i+(j-i)//2
        if(arr1[mid] == x):
            return mid
        elif arr1[mid]<x:
            return BinarySearchElement(arr1,mid+1,j,x) # or use i=mid+1 , keep it in mind, use while loop there
        else:
            return BinarySearchElement(arr1,i,mid-1,x) # or use j=mid-1 ,  keep it in mind, use while loop there
    return -1

## Driver Code
arr = [1,32,5,3,-6,34,-50,10,1,-8,12,'inf','inf','inf','inf']
arr1 = []
for i in arr:
    arr1.append(float(i))
x = float('inf')
i = 0
j = len(arr)-1
result = BinarySearchElement(arr1,i,j,x)
print(result)
```

```python
# Find the First infinity index in an array
# arr = [float('inf'), float('inf'), float('inf'), float('inf'), float('inf')]
arr = [1,2,3,4,5,6,float('inf'),float('inf')]
low = 0
high = len(arr) - 1
ans = -1
while low <= high:
    mid = (low + high) // 2
    if arr[mid] == float('inf'):
        ans = mid
        high = mid - 1
    else:
        low = mid + 1

print(ans)
```
