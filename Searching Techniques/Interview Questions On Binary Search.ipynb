```python
# Find the First infinity index in an array
# arr = [float('inf'), float('inf'), float('inf'), float('inf'), float('inf')]
arr = [1,2,3,4,5,6,float('inf'),float('inf')]
low = 0
high = len(arr) - 1
ans = -1
while low <= high:
    mid = (low + high) // 2

    if arr[mid] == float('inf'):
        ans = mid
        high = mid - 1
    else:
        low = mid + 1

print(ans)
```
