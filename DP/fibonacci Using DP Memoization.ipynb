```python
# Memoization - top down approach - time compexity O(n), space complexity O(n)
def fib_topDown(n,memo):
    if memo[n] is not None:
        return memo[n]
    if n==1 or n==2:
        result = 1
    else:
        result = fib_topDown(n-1,memo)+fib_topDown(n-2,memo)
    memo[n] = result
    return result 
def fib_memo(n):
    memo = [None] * (n+1)
    return fib_topDown(n,memo)
```
