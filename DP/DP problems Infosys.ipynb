### 1
You manage a greenhouse over N time period 
For each period i must choose one of the three climate control mode to maintain stable condition each model has associated energy costs and operational constraints 
Heat :cost H[i]  energy unit
Cool costs C [i]  energy unit 
Idle cost 0 energy unit 

Operational constraints 
Thermal inertia you cannot switch directly bw heat and cool modes you must have atleast one idle period when transitioning from Heat to cool and vice versa 
Active control you cannot remain in idle mode for mare than 2 consecutive periods as the plants require intervention 
Find the minimum total energy cost required to manage the greenhouse across all N periods whole satisfying all constraints 
Input format
The first line contain. Integer N denoting total number of time periods
Each line I of the N  subsequent lines contain integer describing H[i] 
Each line of the N subsequent line contain integer describing C[i]

```python
def greenhouse_memo(N, H, C):
    memo = {}
    def solve(i, prev, idle):
        if i == N:
            return 0
        if (i, prev, idle) in memo:
            return memo[(i, prev, idle)]
        ans = float('inf')
        # Heat (0) → not allowed after Cool (1)
        if prev != 1:
            ans = min(ans, H[i] + solve(i + 1, 0, 0))
        # Cool (1) → not allowed after Heat (0)
        if prev != 0:
            ans = min(ans, C[i] + solve(i + 1, 1, 0))
        # Idle (2) → max 2 consecutive
        if idle < 2:
            ans = min(ans, solve(i + 1, 2, idle + 1))
        memo[(i, prev, idle)] = ans
        return ans
    return solve(0, 2, 0)  # start with idle
# Example
# N = 3
# H = [10,10,10]
# C = [50,50,50]
N = 5
H = [10, 10, 10, 10, 10]
C = [10, 10, 10, 10, 10]
print(greenhouse_memo(N, H, C))  # Output: 10
```

```python
def greenhouse(N, H, C):
    INF = float('inf')
    # States: [Heat, Cool, Idle1, Idle2]
    prev = [H[0], C[0], 0, INF]
    for i in range(1, N):
        curr = [INF] * 4
        # Heat
        curr[0] = min(prev[0], prev[2], prev[3]) + H[i]
        # Cool
        curr[1] = min(prev[1], prev[2], prev[3]) + C[i]
        # Idle (1st)
        curr[2] = min(prev[0], prev[1])
        # Idle (2nd)
        curr[3] = prev[2]
        prev = curr
    return min(prev)
# Inputs
# N = 3
# H = [10, 10, 10]
# C = [50, 50, 50]
N = 5
H = [10, 10, 10, 10, 10]
C = [10, 10, 10, 10, 10]
print(greenhouse(N, H, C))
```
