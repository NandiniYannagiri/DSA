```python
## Implementation of SelectionProcedure
## Recurrence relation T(n) = n+T(m-p) or T(n) = n+T(q-m)
## Time complexity for best case Scenario T(n) = T(n/2)+n --> O(nlogn)
## Time complexity for worst case Scenario T(n) = T(n-1)+n --> O(n^2)
## Space Complexity ---> O(1)
## Function Definition for Partition Alogorithm --> O(n)
def Partition(arr,p,q):
    i=p
    ## starting element as pivot element
    pivot = arr[p]
    for j in range(i+1,len(arr)):
        if arr[j]<=pivot:
            i = i+1
            arr[i],arr[j]=arr[j],arr[i]
    ## Final swap between arr[i] and pivot element
    arr[i],arr[p]=arr[p],arr[i]
    return i+1
## Function Definition
def SelectionProcedure(arr,p,q,k):
    if len(arr)==1:
        return arr[p]
    else:
        m = Partition(arr,p,q)
        if m == k:
            return arr[m-1]
        elif m<k:
            ## Move to right side array
            return SelectionProcedure(arr,m,q,k)
        else:
            ## Move to left side of the array
            return SelectionProcedure(arr,p,m-2,k)
        
arr = [12,100,43,30,9,90]
## arr elements after sorting - [9,12,30,43,90,100]
##                  Positions - [1,2,3,4,5,6]
p = 0
q = len(arr)-1
## Kth postion
k = 2
## Function Call
result = SelectionProcedure(arr,p,q,k)
print("kth smallest element",result)
```
