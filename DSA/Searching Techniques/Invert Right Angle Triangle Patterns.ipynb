```python
## 1 row - 4 stars
## 2 row - 3 stars
## 3 row - 2 stars
## 4 row - 1 star

n = 4
for i in range(1,n+1):
    for j in range(0,n-i):
        print(" ",end=" ")
    for m in range(n-i+1,n+1):
        print("*",end=" ")
    print()
```
