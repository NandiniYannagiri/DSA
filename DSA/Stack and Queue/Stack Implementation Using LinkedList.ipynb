```python
## Implentation using class
class stack:
    def __init__(self):
        self.data = deque()
    # these push,pop,peek are builtin functions in deque
    # we can give another names by our own
    def push(self,data): # push the data into stack
        self.data.append(data)
    def pop(self): # delete the last data from stack
        return self.data.pop()
    def peek(self): # it just give the last data 
        return self.data[-1]
    def sizeOfQueue(self): #
        return len(self.data)
    def isEmpty(self):
        return len(self.data)==0
s = stack()
s.push(10)
s.push(20)
s.push(30)
s.push(40)
s.peek()
```
