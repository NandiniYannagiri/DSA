```python
## Queue Implementation Using LinkedList
class Queue:
    def __init__(self):
        self.buffer = deque()
    def enqueue(self,data):
        self.buffer.appendleft(data)
    def isEmpty(self):
        return len(self.buffer)==0
    def size(self):
        return len(self.buffer)
    def deque(self):
        return self.buffer.pop()
Q = Queue()
Q.enqueue(10)
Q.enqueue(80)
Q.enqueue(70)
Q.enqueue(40)
```
