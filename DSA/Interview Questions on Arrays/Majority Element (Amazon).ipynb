```python
## Majority Element - in this problems they have given if frequency > n/2 then consider it as majority element
## Approach one - sort the array and count - O(nlogn) & space complexity O(n) because extra space is using as counter dictionary
## Approach two - using dictionary(hash data structure)-Key,value
from collections import Counter
def MajorityElement(nums):
    counts = Counter(nums)
    print(counts)
    return max(counts.keys(),key=counts.get)
nums = [2,2,1,1,2,2]
result = MajorityElement(nums)
print(result)
```

```python
## My way
## time complexity ---> O(n)+O(n) --O(n)
## space compexity --->O(n) -- extra space is using for freq dictionary
def MajorityElement(nums):
    freq = {}
    for i in nums:
        if i in freq:
            freq[i] +=1
        else:
            freq[i] = 1
    print(freq)
    most_repeated = 0
    max_count = 0
    for i in freq:
        if freq[i]>max_count:
            max_count = freq[i]
            most_repeated = i
    return most_repeated
                

nums = [2,2,1,1,2,2]
result = MajorityElement(nums)
print(result)
```

```python
## Optimized approach --> Boyer Moore Voting Algorithm -- frequency of element > n/2
def findcandidate(nums):
    count = 0
    candidate = None
    for num in nums:
        if count == 0:
            candidate = num
        count+=(1 if num==candidate else -1)
    
def isMajority(nums,candidate):
    cnt = 0
    n = len(nums)
    for i in range(n):
        if nums[i]==candidate:
            cnt+=1
    if cnt > n/2:
        return 1
    else:
        return 0
    
def MajorityElement(nums):
    cand = findcandidate(nums)
    if isMajority(nums,cand):
        print("Majority element is:" ,cand)
    else:
        print("no majority element")

nums = [2,2,1,1,2,2]
result = MajorityElement(nums)
print(result)
```
