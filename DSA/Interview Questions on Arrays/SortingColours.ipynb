```python
## Sorting Colours
## red-0,green-1,yellow-2
## [1,1,0,2,0,2]
## Time complexity O(n)
## we can aslo do this using merge sort, quicksort.
## but time complexity of both are O(nlogn)
def SortColors(colours):
    p0 = current = 0
    p2 = len(colours)-1
    while current <= p2:
        if colours[current] == 0:
            colours[p0],colours[current] = colours[current],colours[p0]
            p0 += 1
            current += 1
        elif colours[current] == 2:
            colours[p2],colours[current] = colours[current],colours[p2]
            p2 -= 1
        else:
            current += 1
    return colours

## Driver code
colours = [1,1,0,2,0,2]
result = SortColors(colours)
print(result)
```
