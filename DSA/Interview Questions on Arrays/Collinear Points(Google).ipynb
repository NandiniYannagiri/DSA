```python
## Collinear Points
## one way
## By calculating slopes, if slopes are same we can say those are collinear points
def isCollinearPoints(x1, x2, x3, y1, y2, y3):
    ## if slopes are same then we can say the points are collinear
    ##((y2 - y1) / (x2 - x1)) == ((y3 - y2) / (x3 - x1)) ----> do cross multiply
    if ((y2 - y1) * (x3 - x1)) == ((y3 - y2) * (x2 - x1)):
        print("given points are collinear points")
    else:
        print("given points are not cllinear points")

x1, x2, x3, y1, y2, y3 = 1, 1, 1, 6, 0, 9
isCollinearPoints(x1, x2, x3, y1, y2, y3)
```

```python
## Other way ---> if the area of triangle is formed using the points we can say those are not collinear
def isCollinearPoints(x1, x2, x3, y1, y2, y3):
    area = 0.5 * (x1 * (y2 - y3) + x2 * (y3 - y1) + x3 * (y1 - y2))
    print(area)
    if area:
        print("given points are not collinear points")
    else:
        print("given points are cllinear points")
        
x1, x2, x3, y1, y2, y3 = 1, 1, 1, 6, 0, 9
isCollinearPoints(x1, x2, x3, y1, y2, y3)
```
