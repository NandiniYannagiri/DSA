```python
## 1
## Best time to Buy & Sell Stocks
## my Way
def MaxProfit(prices):
    minPrice = min(prices)
    maxPrice = max(prices)
    maxProfit = maxPrice-minPrice
    return maxProfit

prices = [7,1,5,3,6,4,14]
max_profrit = MaxProfit(prices)
print(max_profrit)


## Other Way without using any function

def MaxProfit(prices):
    min_price = float('inf')
    max_profit = 0
    for i in range(len(prices)):
        if prices[i]<min_price:
            min_price = prices[i]
        elif prices[i]-min_price>max_profit:
            max_profit = prices[i]-min_price
    return max_profit


prices = [7,1,5,3,6,4,14]
max_profrit = MaxProfit(prices)
print(max_profrit)
```
