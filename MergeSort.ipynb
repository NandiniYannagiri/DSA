```python
# Merge Sort
# Merge Procedure
def MergeProcedure(arr,i,mid,j):
    # no.of elements in the left subarray
    n1 = (mid-i)+1
    # no.of elements in the right subarray
    n2 = j-(mid+1)+1
    leftsubarray = [0]*n1
    rightsubarray = [0]*n2
    #copy elements from array to subarrays
    for m in range(n1):
        leftsubarray[m] = arr[i+m]
    for n in range(n2):
        rightsubarray[n] = arr[mid+1+n]
    # now, combine the subarrays into single array ----> merging
    p = 0
    q = 0
    k = i
    # compare the values of both sub arrays
    while p<n1 and q<n2:
        if leftsubarray[p]<=rightsubarray[q]:
            arr[k] = leftsubarray[p]
            p+=1
        else:
            arr[k] = rightsubarray[q]
            q+=1
        k+=1
    # copy the entire elements from left subarray
    while p<n1:
        arr[k] = leftsubarray[p]
        p+=1
        k+=1
    # copy the entire elements from right subarray
    while q<n2:
        arr[k] = rightsubarray[q]
        q+=1
        k+=1
def MergeSort(arr,i,j):
    if i<j:
        mid = i+(j-i)//2
        # Recursive call left subtree
        MergeSort(arr,i,mid)
        # Recursive call right subtree
        MergeSort(arr,mid+1,j)
        # Merge bothe subtrees - call merge procedure
        MergeProcedure(arr,i,mid,j)
    return arr

arr = [50,52,10,4,22,33,2,4,7,100]
i = 0
j = len(arr)-1
result = MergeSort(arr,i,j)
print(result)
```
