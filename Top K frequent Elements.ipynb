```python
## one way
from collections import Counter
import heapq
def topKElements(arr,k):
    if k==len(arr):
        return arr
    count = Counter(arr)
    return heapq.nlargest(k,count.keys(),key = count.get)

arr = [1,1,1,2,1,2,3]
k = 2
result = topKElements(arr,k)
print(result)
```

```python
def topKElements(elements, k):
    l = []
    frequency = {}

    if k == len(elements):
        return elements

    # Count frequency
    for ele in elements:
        if ele in frequency:
            frequency[ele] += 1
        else:
            frequency[ele] = 1

    # Sort based on frequency dictionary (NOT d)
    keys = sorted(frequency, key=frequency.get, reverse=True)

    # Take first k keys
    for i in range(k):
        l.append(keys[i])

    return l


# Driver Code
elements = [1,1,1,2,1,2,3]
k = 2
result = topKElements(elements, k)
print(result)
```
