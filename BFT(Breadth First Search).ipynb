```python
from collections import deque
Q = deque()
# finction definition
def BFT(visited,graph,node):
    visited.add(node)
    Q.appendleft(node)
    while Q:
        node = Q.pop()
        print(node,end=" ")
        for adjacentNodes in graph[node]:
            if adjacentNodes not in visited:
                visited.add(adjacentNodes)
                Q.appendleft(adjacentNodes)
                
# Driver Code
visited = set()
graph = {'A' : ['B','C','D'],
         'B' : ['E'],
         'C' : ['E','F'],
         'D' : ['F'],
         'E' : ['G'],
         'F' : ['G'],
         'G' : []
        }
# Function calling
BFT(visited,graph,'A')
```
