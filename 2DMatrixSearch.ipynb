```python
## Matrix Element searching
def Search2DElementMatrrix(Matrix,target):
    m = len(Matrix)#  --> rows
    if m==0:
        return False
    n = len(Matrix[0])# ---> columns
    left, right = 0,m*n-1
    while left<=right:
        mid = left + (right-left) // 2
        ## finding the indeces of row and column
        ## row = mid//n, column = mid%n, here n = no.of columns
        mid_element = Matrix[mid//n][mid%n]
        if target == mid_element:
            return True
        elif target<mid_element:
            right = mid-1
        else:
            left = mid+1
    return False

Matrix = [[1,3,5,7],[10,11,16,20],[23,30,34,60]] # -----> row wise matrix, means, take matrix element as 1 row
target = 16
result = Search2DElementMatrrix(Matrix,target)
print(result)
```
