```python
## Reversal of LinkedList
class Node:
    def __init__(self,data):
        self.data = data
        self.next = None
class LinkedList:
    def __init__(self):
        self.head = None
    def insertionAtEnd(self,new_data):
        new_node = Node(new_data)
        if self.head is None:
            self.head = new_node
            return
        else:
            temp = self.head
            while temp.next:
                temp = temp.next
            temp.next = new_node
    def ReverseLinkedList(self):
        prev = None
        next_prev = None
        curr = self.head
        while curr:
            next_ptr = curr.next
            curr.next = prev
            prev = curr
            curr = next_ptr
        self.head = prev
    def printList(self):
        temp = self.head
        while temp:
            print(f"{temp.data}",end = " ")
            temp = temp.next
LList = LinkedList()
LList.insertionAtEnd(10)
LList.insertionAtEnd(20)
LList.insertionAtEnd(30)
LList.insertionAtEnd(40)
LList.insertionAtEnd(50)
LList.printList()
LList.ReverseLinkedList()
print()
LList.printList()
            
                
                
```
