```python
class Node:
    def __init__(self,data):
        self.data = data
        self.next = None
class LinkedList:
    def __init__(self):
        self.head = None
    def insertionAtEnd(self,new_data):
        new_node = Node(new_data)
        if self.head is None:
            self.head = new_node
        else:
            temp = self.head
            while temp.next:
                temp = temp.next
            temp.next = new_node
    def MergingLinkedLists(self,head1,head2):
        temp = None
        if head1 is None:
            return head2
        elif head2 is None:
            return head1
        elif head1.data<=head2.data:
            temp = head1
            temp.next = self.MergingLinkedLists(head1.next,head2)
        else:
            temp = head2
            temp.next = self.MergingLinkedLists(head1,head2.next)
        return temp
    def printList(self):
        temp = self.head
        while temp:
            print(str(temp.data)+" ",end=" ")
            temp=temp.next
Llist1 = LinkedList()
Llist1.insertionAtEnd(1)
Llist1.insertionAtEnd(3)
Llist1.insertionAtEnd(5)
Llist1.printList()
print()
Llist2 = LinkedList()
Llist2.insertionAtEnd(2)
Llist2.insertionAtEnd(4)
Llist2.insertionAtEnd(6)
Llist2.printList()
print()
Llist3 = LinkedList()
Llist3.head = Llist3.MergingLinkedLists(Llist1.head,Llist2.head)
Llist3.printList()
```
