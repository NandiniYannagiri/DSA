```python
## Searching in LinkedList
## Time complexity is O(n) - because need to traverse the entire Linked List
## there is no access for random access so, Binary search can't be implemented Like array.
## so traverse the entire LinkedList
## but, we can use rondomization concept - skip list ---> Advanced DSA
class Node:
    def __init__(self,data):
        self.data = data
        self.next = None
class LinkedList:
    def __init__(self):
        self.head = None
    def InsertionatEnd(self,new_data):
        new_node = Node(new_data)
        if self.head == None:
            self.head = new_node
            return
        else:
            temp = self.head
            while temp.next:
                temp = temp.next
            temp.next = new_node
    def searchData(self,key):
        temp = self.head
        while temp:
            if temp.data == key:
                print(f"given value {key} is present in LinkedList")
                return
            temp = temp.next
        print(f"given value {key} is not present in LinkedList")
        return#False
    def printList(self):
        temp = self.head
        while temp:
            print(f"{temp.data} ",end = " ")
            temp = temp.next
LList = LinkedList()
LList.InsertionatEnd(10)
LList.InsertionatEnd(20)
LList.InsertionatEnd(43)
LList.InsertionatEnd(57)
LList.InsertionatEnd(100)
LList.printList()
print()
LList.searchData(100)
LList.searchData(22)
```
