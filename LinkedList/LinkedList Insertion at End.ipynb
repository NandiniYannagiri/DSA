```python
class Node:
    def __init__(self,data):
        self.data = data
        self.next = None
class LinkedList:
    def __init__(self):
        self.head = None
    ## insertion at end
    def InsertionAtEnd(self,new_data):
        new_node = Node(new_data)
        if self.head is None:
            self.head = new_node
            return
        temp = self.head
        while temp.next:
            temp = temp.next
        temp.next = new_node
    def printList(self):
        temp = self.head
        while temp:
            print(str(temp.data)+" ", end=" ")
            temp =temp.next
    def countingNodes(self):
        temp = self.head
        count = 0
        while temp:
            count+=1
            temp = temp.next
        print(count)
LList = LinkedList()
LList.InsertionAtEnd(2)
LList.InsertionAtEnd(4)
LList.InsertionAtEnd(5)
LList.InsertionAtEnd(10)
LList.InsertionAtEnd(5)
LList.InsertionAtEnd(10)
LList.printList()
print()
LList.countingNodes()
```
