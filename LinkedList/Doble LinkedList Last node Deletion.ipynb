```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None
        self.prev = None

class DoublyLinkedList:
    def __init__(self):
        self.head = None
    # Insert at End
    def InsertionAtEnd(self, new_data):
        new_node = Node(new_data)
        if self.head is None:
            self.head = new_node
            return
        temp = self.head
        while temp.next:
            temp = temp.next
        temp.next = new_node
        new_node.prev = temp

    def DeleteLastNode(self):
        if self.head is None:
            print("List is empty")
            return
        # If only one node
        if self.head.next is None:
            self.head = None
            return
        temp = self.head
        # Move to last node
        while temp.next:
            temp = temp.next
        # Remove last node
        temp.prev.next = None
    # Print List
    def printList(self):
        temp = self.head
        while temp:
            print(temp.data, end=" ")
            temp = temp.next
        print()
```

```python
dll = DoublyLinkedList()

dll.InsertionAtEnd(2)
dll.InsertionAtEnd(4)
dll.InsertionAtEnd(5)
dll.InsertionAtEnd(10)

print("Original List:")
dll.printList()

dll.DeleteLastNode()

print("After Deleting Last Node:")
dll.printList()
```
