```python
class Node:
    def __init__(self,data):
        self.data = data
        self.next = None
        self.prev = None
class DoubleLinkedList:
    def __init__(self):
        self.head = None
    def insertionAtBeginning(self,new_data):
        new_node = Node(new_data)
        if self.head is None: 
            self.head = new_node
        else:
            new_node.next = self.head
            self.head.prev = new_node
            self.head = new_node
    def insertionAtEnd(self,new_data):
        new_node = Node(new_data)
        if self.head is None: 
            self.head = new_node
        else:
            temp = self.head
            while temp.next:
                temp = temp.next
            temp.next = new_node
            new_node.prev = temp
    def Display(self):
        temp = self.head
        while temp:
            print(temp.data,end=" ")
            temp = temp.next
DLL = DoubleLinkedList()
DLL.insertionAtBeginning(2)
DLL.insertionAtBeginning(1)
DLL.insertionAtBeginning(3)
DLL.insertionAtEnd(40)
DLL.insertionAtEnd(50)
DLL.Display()
        

# insertionAtBeginning(10)
```

```python
# Stack Implementation Using Linked List
class Node:
    def __init__(self,data):
        self.data = data
        self.next = None
class Stack:
    def __init__(self):
        self.head = None
    def push(self,new_data):
        new_node = Node(new_data)
        if self.head is None:
            self.head = new_node
        else:
            new_node.next = self.head
            self.head = new_node
    def delete(self):
        if self.isEmpty():
            print("Linked List is Empty ,Can not do the deletion")
        else:
            print(self.head.data)
            self.head = self.head.next
    def peek(self):
        if self.isEmpty():
            return
        return self.head.data
    
    def isEmpty(self):
        if self.head is None:
            return True
        else:
            return False
    def printList(self):
        temp = self.head
        while temp:
            print(temp.data,end = " ")
            temp = temp.next
stack = Stack()
stack.isEmpty()
stack.push(10)
stack.push(20)
stack.push(30)
stack.printList()
print()
stack.delete()
stack.printList()
print()
stack.delete()
stack.printList()
print()
stack.delete()
stack.printList()
stack.delete()
```
