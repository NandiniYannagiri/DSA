```python
import heapq
# Create a list and transform it into a heap
heap = [5, 7, 9, 1, 3]
heapq.heapify(heap)
print("Heap:", heap) # Output: [1, 3, 9, 7, 5]
# Push a new item onto the heap
heapq.heappush(heap, 4)
print("Heap after push:", heap) # Output: [1, 3, 4, 7, 5, 9]
# Pop the smallest item from the heap
smallest = heapq.heappop(heap)
print(smallest)
print(heapq.heappop(heap))
print(heapq.heappop(heap))
print(heapq.heappop(heap))
print(heapq.heappop(heap))
print(heapq.heappop(heap))
print(heapq.heappop(heap))
# print("Smallest element:", smallest) # Output: 1
# print("Heap after pop:", heap) # Output: [3, 5, 4, 7, 9]
```

```python
import heapq
# Create a list and transform it into a heap
values = [5, 1, 3, 7, 4, 2]
heapq.heapify(values)
print("Heap:", values) # Output: [1, 4, 2, 7, 5, 3]
# Get the 3 smallest elements
n_smallest = heapq.nsmallest(3, values)
print("Smallest 3 elements:", n_smallest) # Output: [1, 2, 3]
# Get the 2 largest elements
n_largest = heapq.nlargest(2, values)
print("Largest 2 elements:", n_largest) # Output: [7, 5]
```

```python
## one way
from collections import Counter
import heapq
def topKElements(arr,k):
    if k==len(arr):
        return arr
    count = Counter(arr)
    print(count)
    return heapq.nlargest(k,count.keys(),key = count.get)

arr = [1,1,1,2,1,2,3]
k = 2
result = topKElements(arr,k)
print(result)
```

```python
def topKElements(elements,k):
    l=[]
    frequency = {}
    if k==len(elements):
        return elements
    for ele in elements:
        if ele in frequency:
            frequency[ele]+=1
        else:
            frequency[ele]=1
    print(frequency)
    # return frequency
    keys = sorted(frequency,key=frequency.get,reverse=True)
    print(keys)
    for i in range(k):
        l.append(keys[i])
    return l


# Driver Code
elements = [1,6,3,3,3,3,1,2,1,2,3,4]
k = 2 # ---->indicating top 2 elements(most repeated)
result = topKElements(elements,k)
print(result)
```
