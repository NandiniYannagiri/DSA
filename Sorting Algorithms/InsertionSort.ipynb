```python
## Function definition
def InsertionSort(arr):
    for i in range(1,len(arr)):
        key = arr[i]
        j = i-1
        while j>=0 and key<arr[j]:
            arr[j+1]=arr[j]
            j-=1
        arr[j+1]=key
    return arr

arr = [12,3,22,5,53,46,21,2,4,7,43]
result = InsertionSort(arr)
print(result)
```
