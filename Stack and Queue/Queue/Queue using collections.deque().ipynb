```python
# Queue using deque()
from collections import deque
q = deque()
# dir(q)
q.append(10)      # Enqueue
q.append(20)
q.append(30)      # Enqueue
q.append(40)
print(q)
q.popleft()       # Dequeue
print(q)
q.popleft()   
print(q)
# print("Front:", q[0])
# q.rotate(2)
# print(q)
```

```python
# Queue using deque()
from collections import deque

q = deque()

q.appendleft(10)      # Enqueue
q.appendleft(20)
# print(q)
q.appendleft(30)      # Enqueue
q.appendleft(40)
print(q)
q.pop()       # Dequeue
print(q)

# print("Front:", q[0])
```
