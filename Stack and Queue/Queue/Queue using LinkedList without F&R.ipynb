```python
# Node class
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None
# Queue using Linked List
class Queue:
    def __init__(self):
        self.head = None
    # Insert at end (Enqueue)
    def enqueue(self, data):
        new_node = Node(data)
        if self.head is None:
            self.head = new_node
            return
        temp = self.head
        while temp.next:
            temp = temp.next
        temp.next = new_node
    # Delete from beginning (Dequeue)
    def dequeue(self):
        if self.head is None:
            print("Queue is Empty")
            return
        temp = self.head
        self.head = self.head.next
        return temp.data
    # Display
    def display(self):
        temp = self.head
        while temp:
            print(temp.data, end=" ")
            temp = temp.next
q = Queue()
q.enqueue(10)
q.enqueue(20)
q.enqueue(30)
q.display()
print("\nRemoved:", q.dequeue())
q.display()
print("\nRemoved:", q.dequeue())
q.display()
```
