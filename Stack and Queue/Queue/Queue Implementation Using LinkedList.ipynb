```python
from collections import deque

class Queue:
    def __init__(self):
        self.adding_data = deque()

    def enqueue(self, data):
        self.adding_data.appendleft(data)

    def isEmpty(self):
        return len(self.adding_data) == 0

    def size(self):
        return len(self.adding_data)

    def dequeue(self):
        if self.isEmpty():
            print("Queue Underflow")
            return None
        return self.adding_data.pop()

    def printList(self):
        if self.isEmpty():
            print("Queue is empty")
            return
        for item in self.adding_data:
            print(item, end=" ")
        print()
Q = Queue()
Q.enqueue(10)
Q.enqueue(80)
Q.enqueue(70)
Q.enqueue(40)
```
