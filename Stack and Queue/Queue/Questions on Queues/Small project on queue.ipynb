```python
from collections import deque

class CallCenter:
    def __init__(self):
        self.queue = deque()
    def add_customer(self, name):
        self.queue.append(name)
        print(f"{name} added to waiting list.")
    def serve_customer(self):
        if not self.queue:
            print("No customers in queue.")
        else:
            served = self.queue.popleft()
            print(f"{served} is being served.")
    def next_customer(self):
        if not self.queue:
            print("No customers waiting.")
        else:
            print("Next customer:", self.queue[0])
    def display_queue(self):
        if not self.queue:
            print("Queue is empty.")
        else:
            print("Waiting Customers:", list(self.queue))
    def total_customers(self):
        print("Total waiting customers:", len(self.queue))

# Main Program
center = CallCenter()
choice = ""
while choice != "6":
    print("\n--- Call Center Menu ---")
    print("1. Add Customer")
    print("2. Serve Customer")
    print("3. View Next Customer")
    print("4. Display Waiting List")
    print("5. Total Customers")
    print("6. Exit")
    choice = input("Enter choice: ")
    if choice == "1":
        name = input("Enter customer name: ")
        center.add_customer(name)
    elif choice == "2":
        center.serve_customer()
    elif choice == "3":
        center.next_customer()
    elif choice == "4":
        center.display_queue()
    elif choice == "5":
        center.total_customers()
    elif choice == "6":
        print("Exiting system...")
    else:
        print("Invalid choice. Try again.")
```
