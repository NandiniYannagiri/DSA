```python
class QueueOneStack:
    def __init__(self):
        self.stack = []
    def enqueue(self, x):
        self.stack.append(x)
    def dequeue(self):
        if not self.stack:
            return "Queue is Empty"
        if len(self.stack) == 1:
            return self.stack.pop()
        temp = self.stack.pop()
        item = self.dequeue()
        self.stack.append(temp)
        return item
    def isEmpty(self):
        return len(self.stack) == 0
q = QueueOneStack()
q.enqueue(10)
q.enqueue(20)
q.enqueue(30)
print(q.dequeue())  # 10
print(q.dequeue())  # 20
# print(q.dequeue())  # 30
# print(q.dequeue())  # Queue is Empty
```
