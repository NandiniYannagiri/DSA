```python
import heapq

# create empty priority queue
pq = []

# insert elements
heapq.heappush(pq, 30)
heapq.heappush(pq, 10)
heapq.heappush(pq, 20)
heapq.heappush(pq, 5)

print("Priority Queue:", pq)

# remove elements (smallest first)
print("Removed:", heapq.heappop(pq))
print("Removed:", heapq.heappop(pq))

print("Remaining Queue:", pq)
```
