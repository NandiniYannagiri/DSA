```python
# Generating Binary Numbers from 1 to N Using Queue
from collections import deque

def generate_binary(n):
    q = deque()
    q.append("1")

    for _ in range(n):
        current = q.popleft()
        print(current)
        q.append(current + "0")
        q.append(current + "1")

# Driver
n = 5
generate_binary(n)
```
