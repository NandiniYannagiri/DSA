```python
# Queue using doubly linked list, but dont need. because, queue has generally only one direction. so sinlge linked list can enough. 
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None
        self.prev = None


class Queue:
    def __init__(self):
        self.front = None
        self.rear = None

    def enqueue(self, data):
        new_node = Node(data)

        if self.rear is None:
            self.front = self.rear = new_node
        else:
            self.rear.next = new_node
            new_node.prev = self.rear
            self.rear = new_node

        print(f"{data} inserted")

    def dequeue(self):
        if self.front is None:
            print("Queue is Empty")
            return

        temp = self.front
        self.front = self.front.next

        if self.front:
            self.front.prev = None
        else:
            self.rear = None

        print(f"{temp.data} removed")

    def display(self):
        temp = self.front
        while temp:
            print(temp.data, end=" <-> ")
            temp = temp.next
        print("None")


# Driver Code
q = Queue()
q.enqueue(10)
q.enqueue(20)
q.enqueue(30)
q.display()
q.dequeue()
q.display()
```
