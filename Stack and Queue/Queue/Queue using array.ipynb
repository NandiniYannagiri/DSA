```python
class Queue:
    def __init__(self):
        self.queue = []

    def enqueue(self, data):
        self.queue.append(data)
        print(f"{data} inserted")
        
    def isEmpty(self):
        return len(self.queue) == 0

    def dequeue(self):
        if self.isEmpty():
            print("Queue is Empty")
        else:
            removed = self.queue.pop(0)
            print(f"{removed} removed")

    def peek(self):
        if self.isEmpty():
            print("Queue is Empty")
        else:
            print("Front element:", self.queue[0])

    
    def display(self):
        print("Queue:", self.queue)


# Driver Code
q = Queue()
q.enqueue(10)
q.enqueue(20)
q.enqueue(30)
q.display()
q.dequeue()
q.display()
```
